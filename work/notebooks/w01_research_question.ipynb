## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

My chosen lane is **Structured Content Archetype Clustering**. Imagine you have a huge pile of different types of content, like articles, videos, or product descriptions. This approach is like having a super-smart sorting hat that looks at all the pieces of content and automatically groups together the ones that are very similar, even if they're not explicitly labeled. We're looking for the 'archetypes' or typical patterns of content that exist.

Why this lane? Well, if we can understand these underlying groups, we can make much better decisions. For example, if we know what kind of content a user likes, we can show them more similar content from the same 'archetype'. Or, if one archetype isn't performing well, we can focus our efforts on improving that *type* of content. It helps us understand our content better and ultimately give users what they want more effectively.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**What decision does this work improve?**

Our project helps us decide *which content to show to whom* and *how to create new content*. By grouping similar content, we can figure out what types of content are popular or unpopular. This helps us make smarter choices about what content to promote, what to improve, and what new content ideas might be successful. Think of it like a music streaming service suggesting new songs based on what genres you already like – our project helps make those kinds of smart suggestions for content.

**Who acts on the output, and what do they do?**

There are a few key players who'd use this:

*   **Content Creators/Editors:** They can see which 'archetypes' of content are doing well and get ideas for new topics or styles that are likely to attract readers. If a certain type of content isn't performing, they know to either fix it or stop making it.
*   **Website Managers/Marketers:** They can use these groups to personalize what users see. If a user enjoys content from a specific archetype, the website can show them more of that kind of content, making their experience better and keeping them on the site longer.

**What does a wrong recommendation cost?**

If our system makes a 'wrong' recommendation (meaning it groups content incorrectly or gives bad advice based on those groups), it could have a few costs:

*   **Wasted Time/Money:** Content creators might spend time making content that doesn't resonate with the audience, because our 'archetypes' were misleading. This means wasted effort and resources.
*   **Unhappy Users:** If users keep seeing content that doesn't match their interests because our system incorrectly categorized it, they might get frustrated and stop visiting the site or using the service.
*   **Missed Opportunities:** We might miss out on discovering popular new content trends if our clustering isn't picking them up correctly, or fail to promote valuable content because it was grouped with less relevant items. These are all reasons why getting this clustering right is important!

```python
from datasets import load_dataset
from google.colab import userdata
import getpass

# Authenticate to Hugging Face
# You can store your Hugging Face token as a Colab Secret named 'HF_TOKEN'
# Or enter it manually when prompted

HF_TOKEN = userdata.get('HF_Token')

if not HF_TOKEN:
    HF_TOKEN = getpass.getpass('Enter your Hugging Face token: ')
```

```python
from datasets import load_dataset
import pandas as pd
import duckdb

print("Loading warehouse dataset...")
dataset = load_dataset(
    'FlyRank/internship-warehouse',
    name='fact_content_daily_performance',
    token=HF_TOKEN
)

train_split = dataset['train']

print("Processing in batches and filtering for March 2026...")
march_rows = []
batch_size = 50000

# Iterate in chunks
for i in range(0, len(train_split), batch_size):
    print(f"  Batch {i // batch_size + 1}...")

    batch = train_split.select(range(i, min(i + batch_size, len(train_split))))
    batch_df = batch.to_pandas()

    # Filter for March 2026
    batch_df['report_date'] = pd.to_datetime(batch_df['report_date'])
    march_batch = batch_df[batch_df['report_date'].dt.to_period('M') == '2026-03']

    march_rows.append(march_batch)
    print(f"    → {len(march_batch):,} March rows")

# Combine all March batches
march_2026 = pd.concat(march_rows, ignore_index=True)

print(f"\n✓ Total March 2026 rows: {len(march_2026):,}\n")

# NOW use DuckDB for analysis
print("--- DuckDB Analysis ---")

stats = duckdb.sql("""
    SELECT
        COUNT(*) as total_rows,
        COUNT(DISTINCT content_hash_id) as unique_content,
        COUNT(DISTINCT client_hash_id) as unique_clients,
        MIN(report_date) as min_date,
        MAX(report_date) as max_date
    FROM march_2026
""").to_df()

print(stats)

# Cluster separability: CTR variance
ctr_stats = duckdb.sql("""
    SELECT
        COUNT(*) as rows_with_ctr,
        MIN(gsc_clicks / NULLIF(gsc_impressions, 0) * 100) as min_ctr_pct,
        MAX(gsc_clicks / NULLIF(gsc_impressions, 0) * 100) as max_ctr_pct,
        AVG(gsc_clicks / NULLIF(gsc_impressions, 0) * 100) as avg_ctr_pct,
        STDDEV(gsc_clicks / NULLIF(gsc_impressions, 0) * 100) as stddev_ctr_pct
    FROM march_2026
    WHERE gsc_impressions > 0
""").to_df()

print("\n--- CTR Variance (Cluster Separability) ---")
print(ctr_stats)

print(f"\n--- First 5 Rows ---")
display(march_2026.head())
```

```python
# Position variance — pages rank in very different spots
position_stats = duckdb.sql("""
    SELECT
        COUNT(*) as rows_with_position,
        MIN(gsc_avg_position) as min_position,
        MAX(gsc_avg_position) as max_position,
        AVG(gsc_avg_position) as avg_position,
        STDDEV(gsc_avg_position) as stddev_position
    FROM march_2026
    WHERE gsc_avg_position > 0  -- Exclude "no data" rows
""").to_df()

print("--- Position Variance (Cluster Separability) ---")
print(position_stats)
```

```python
# Traffic source diversity — different content attracts different channels
traffic_mix = duckdb.sql("""
    SELECT
        COUNT(*) as rows_with_sessions,
        COUNT(DISTINCT client_hash_id) as clients,
        AVG(sessions_organic) as avg_organic,
        AVG(sessions_direct) as avg_direct,
        AVG(sessions_ai) as avg_ai,
        COUNT(CASE WHEN sessions_ai > 0 THEN 1 END) as rows_with_ai
    FROM march_2026
    WHERE (sessions_organic + sessions_direct + sessions_ai) > 0
""").to_df()

print("--- Traffic Source Mix ---")
print(traffic_mix)
```

```python
# Pages with SIMILAR CTR but DIFFERENT position tiers
# This proves that grouping by CTR alone misses important distinctions

similar_ctr_diff_position = duckdb.sql("""
    SELECT
        FLOOR((gsc_clicks / NULLIF(gsc_impressions, 0) * 100) / 0.5) * 0.5 as ctr_bucket_pct,
        COUNT(*) as page_count,
        AVG(gsc_avg_position) as avg_position,
        STDDEV(gsc_avg_position) as stddev_position,
        MIN(gsc_avg_position) as min_position,
        MAX(gsc_avg_position) as max_position,
        AVG(ga4_engaged_sessions) as avg_engagement,
        STDDEV(ga4_engaged_sessions) as stddev_engagement
    FROM march_2026
    WHERE gsc_impressions > 100  -- Minimum volume to avoid noise
        AND gsc_avg_position > 0
        AND ga4_engaged_sessions >= 0
    GROUP BY ctr_bucket_pct
    ORDER BY ctr_bucket_pct
    LIMIT 15
""").to_df()

print("--- Pages with SAME CTR but DIFFERENT Positions/Engagement ---")
print("(This shows CTR alone doesn't define archetypes)")
print(similar_ctr_diff_position)
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This study can say:

* We found several groups (called clusters or archetypes) by analyzing 331,000 pages using data like CTR, search ranking, user engagement, and traffic sources.

* Pages in the same group have similar performance patterns. For example, one group may have high CTR and high rankings, while another may have low CTR and low rankings.

* If we run the same analysis again on the same data, we get the same groups, showing the results are consistent.

* Since pages in the same group behave similarly, they may need similar types of improvements. For example:

  * High-performing pages → monitor and maintain.
  * Low-performing pages → rewrite or improve.
  * Old or weak pages → consider combining or removing.

* These groups help editors organize and prioritize 331,000 pages instead of reviewing each page one by one.

* Each group includes an explanation of why the pages were grouped together, such as low CTR with low rankings or high traffic but declining rankings.


This study cannot say:

* Improving a page will definitely increase its CTR. The study only finds patterns—it does not prove what works.

* A page performs poorly because of its search ranking. The data shows they are related, but not that one causes the other.

* The clusters can predict what Google will do next or reveal how Google's ranking algorithm works.

* The clusters are based on the meaning or topic of the content. They are based only on performance data, not on what the pages are about.

* A cluster proves that a page is high quality. High quality is subjective and was not measured in this study.

* The clusters are the only correct way to group pages. Different methods could produce different groups.

* Following the suggested actions is guaranteed to improve performance. Those actions are recommendations based on patterns, not proven results.

* These same groups will work for every website or company. Different websites may have different patterns, so the results may not always apply.
